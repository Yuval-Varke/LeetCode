1. Factorial

```python
def fact(n):
    if n == 0 or n == 1:
        return 1
    return n * fact(n-1)
```

```python
fact(3)
```

```python
fact(4)
```

2. Fibonacci 

```python
def fib(n):
    if n == 0:
        return 0
    elif n == 1:
        return 1
    else:
        return fib(n-1) + fib(n-2)
```

```python
fib(5)
```

```python
fib(10)
```

```python
fib(1)
```
