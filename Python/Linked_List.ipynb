## Linked List

### 1. Detect cycle in Linked list 

```python
class ListNode:
    def __init__(self, val=0, next=None):
        self.val = val
        self.next = next
```

```python
class Solution:
    def has_cycle(self,head):
        slow, fast = head, head
        while fast and fast.next:
            slow = slow.next
            fast = fast.next.next
            if slow == fast:
                return True
        return False

sol = Solution()
```

```python
# Test Case 1 - Linked List with a Cycle

# create nodes
n1 = ListNode(1)
n2 = ListNode(2)
n3 = ListNode(3)
n4 = ListNode(4)

# Link nodes
n1.next = n2
n2.next = n3
n3.next = n4
n4.next = n2
```

```python
# Test 1

sol.has_cycle(n1)
```

```python
# Test Case 2 - Linked List without a Cycle

# Create nodes
a1 = ListNode(1)
a2 = ListNode(2)
a3 = ListNode(3)
a4 = ListNode(4)

# Link nodes
a1.next = a2
a2.next = a3
a3.next = a4
```

```python
# Test 2

sol.has_cycle(a1)
```

```python
# Test Case 3 - Single Node

single = ListNode(1)
```

```python
# Test 3

sol.has_cycle(single)
```

```python
# Test Case 4 - Single Node self-cycle

self_cycle = ListNode(1)
self_cycle.next = self_cycle
```

```python
# Test 4

sol.has_cycle(self_cycle)
```
