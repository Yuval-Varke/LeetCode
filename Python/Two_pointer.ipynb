### 1. Remove Duplicates from sorted array 

. Given a sorted array, remove duplicates in place and return the length of the new array
containing unique elements

· Input: [1,1,2,2,3]

· Output: 3

```python
def remove_dup(arr):
    i = 0
    j = i + 1
    while j < len(arr):
        if arr[i] == arr[j]:
            j += 1
        else:
            arr[i+1],arr[j] = arr[j],arr[i+1]
            i = i+1
            j += 1
    return i+1
```

```python
remove_dup([1,1,2,2,3])
```

```python
remove_dup([1,1,1,1,2,2,2,3,4,5,7])
```
