### 1. Maximum depth of a Binary Tree

Given the root of a binary tree, return its maximum depth

```python
class TreeNode:
    def __init__(self, val):
        self.val = val
        self.left = None
        self.right = None
```

```python
root = TreeNode(40)
root.left = TreeNode(30)
root.right = TreeNode(50)

root.left.left = TreeNode(25)
root.left.right = TreeNode(35)
root.right.left = TreeNode(45)
root.right.right = TreeNode(60)
```

```python
def max_depth(node):
    if not node:
        return 0

    left = max_depth(node.left)
    right = max_depth(node.right)
    return 1 + max(left,right)
```

```python
max_depth(root)
```

### 2. Count Nodes in a Tree

```python
from collections import deque

def count_nodes(node):
    if not node:
        return 0
    q = deque([node])
    cnt = 0
    while q:
        curr = q.popleft()
        cnt += 1
        if curr.left:
                q.append(curr.left)
        if curr.right:
                q.append(curr.right)
    print(cnt)
```

```python
count_nodes(root)
```

```python
def count_nodes_recursive(node):
    if not node:
        return 0
    return 1 + count_nodes_recursive(node.left) + count_nodes_recursive(node.right)
```

```python
count_nodes_recursive(root)
```

### 3. Sum of Nodes in Binary Tree

```python
def sum_nodes(node):
    if not node:
        return 0
    q = deque([node])
    sum = 0
    while q:
        curr = q.popleft()
        sum += curr.val
        if curr.left:
                q.append(curr.left)
        if curr.right:
                q.append(curr.right)
    print(sum)
```

```python
sum_nodes(root)
```

```python
def sum_nodes_recursive(node):
    if not node:
        return 0
    return node.val + sum_nodes_recursive(node.left) + sum_nodes_recursive(node.right)
```

```python
sum_nodes_recursive(root)
```

### 4. Validate BST

```python
import math

def validate_bst(root):
    def traverse(node,low,high):
        if not node:
            return True

        if not (low < node.val < high):
            return False

        return (
            traverse(node.left, low, node.val) and
            traverse(node.right, node.val, high)
        )
    return traverse(root, -math.inf, math.inf)
```

```python
validate_bst(root)
```

### 5. Lowest Common Ancestor in a BST

Given a BST and two nodes p and q, find their lowest common ancestor

```python
root = TreeNode(6)

root.left = TreeNode(2)
root.right = TreeNode(8)

root.left.left = TreeNode(0)
root.left.right = TreeNode(4)

root.right.left = TreeNode(7)
root.right.right = TreeNode(9)

root.left.right.left = TreeNode(3)
root.left.right.right = TreeNode(5)
```

```python
def lowest_common_ancestor(root,p,q):
    while(root):
        if p.val < root.val and q.val < root.val:
            root = root.left
        elif p.val > root.val and q.val > root.val:
            root = root.right
        else:
            return root
```

```python
lowest_common_ancestor(root,p=root.left,q=root.right).val
```

```python
lowest_common_ancestor(root,p=root.left,q=root.left.right).val
```

### 6. K-th smallest element in a BST

```python
root = TreeNode(40)

root.left = TreeNode(30)
root.right = TreeNode(50)

root.left.left = TreeNode(25)
root.left.right = TreeNode(35)

root.right.left = TreeNode(45)
root.right.right = TreeNode(60)
```

```python
def kth_smallest(node,k):
    stack = []

    while True:
        while node:
            stack.append(node)
            node = node.left
        node = stack.pop()
        k -= 1

        if k == 0:
            return node.val

        node = node.right
```

```python
kth_smallest(root,3)
```

```python
kth_smallest(root,1)
```

### 7. Maximum value in N-ary Tree

```python
class NAryNode:
    def __init__(self, val):
        self.val = val
        self.children = []
```

```python
root = NAryNode(10)

#level 1
node_20 = NAryNode(20)
node_30 = NAryNode(30)
node_40 = NAryNode(40)
root.children = [node_20, node_30, node_40]

#level 2
node_20.children = [NAryNode(50), NAryNode(60), NAryNode(70)]
node_30.children = [NAryNode(80)]

node_40.children = [NAryNode(90), NAryNode(100)]
```

```python
import math

def max_nary(node):
    if not node:
        return -math.inf
    max_val = node.val
    for child in node.children:
        max_val = max(max_val,max_nary(child))
    return max_val
        
```

```python
max_nary(root)
```

### 8. Height of N ary Tree

```python
def height_nary(node):
    if not node:
        return 0
    heights = []
    for child in node.children:
        heights.append(height_nary(child))
    return 1 + max(heights,default=0)
```

```python
height_nary(root)
```

### 9. Maximun Path Sum in a Binary Tree

1.Find the maximum path sum in a binary tree

2.A path can start and end at any node, but must go downwards through parent-child connections

```python
root = TreeNode(-10)

root.left = TreeNode(9)
root.right = TreeNode(20)

root.right.left = TreeNode(15)
root.right.right = TreeNode(7)
```

```python
import math

def max_path_sum(node):
    max_sum = -math.inf

    def helper(node):
        nonlocal max_sum

        if not node:
            return 0

        left_gain = max(helper(node.left), 0)
        right_gain = max(helper(node.right), 0)

        current_path = node.val + left_gain + right_gain
        max_sum = max(max_sum, current_path)

        return node.val + max(left_gain, right_gain)

    helper(node)
    return max_sum
```

```python
max_path_sum(root)
```

### 10. Sum of all left leaf nodes

```python
root = TreeNode(3)

root.left = TreeNode(9)
root.right = TreeNode(20)

root.right.left = TreeNode(15)
root.right.right = TreeNode(7)
```

```python
def left_leaf_sum(node):
    if not node:
        return 0

    total = 0
    if node.left and not node.left.left and not node.left.right:
        total += node.left.val

    total += left_leaf_sum(node.left)
    total += left_leaf_sum(node.right)

    return total
```

```python
left_leaf_sum(root)
```
