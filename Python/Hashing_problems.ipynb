### 1. Top K frequent elements

```python
from collections import Counter
```

```python
def top_freq(arr,k):
    freq = Counter(arr)
    return sorted(freq,key=lambda x: freq[x], reverse = True)[:k]
```

```python
arr = [1,2,3,4,2,5,3,45,2,45,2,45,23,1,1,23,34,12,12,2]
```

```python
top_freq(arr,5)
```

```python
top_freq(arr,2)
```

### 2. Valid Anagrams

```python
def is_Anagram(s1,s2):
    if len(s1) != len(s2):
        return False

    return Counter(s1) == Counter(s2)
```

```python
inputs = [
('thing', 'night'),
('loving', 'caring'),
('listen', 'silent'),
('inch', 'chin'),
('hand', 'band')
]
```

```python
for inp in inputs:
    print(f'{inp[0]}, {inp[1]}: {is_Anagram(inp[0], inp[1])}\n')
```

### 3. Isomorphic strings

```python
def Is_Isomorphic(s1,s2):
    if len(s1) != len(s2):
        return False

    map1 = dict()
    map2 = dict()

    for c1,c2 in zip(s1,s2):
        if c1 in map1 and map1[c1] != c2:
            return False
        if c2 in map2 and map2[c2] != c1:
            return False

        map1[c1] = c2
        map2[c2] = c1
        
    return True
```

```python
inputs = [
    ('egg', 'add'),
    ('foo', 'bar'),
    ('paper', 'title'),
    ('ab', 'aa')
]
```

```python
for inp in inputs:
    print(f'{inp[0]}, {inp[1]}: {Is_Isomorphic(inp[0], inp[1])}\n')
```
