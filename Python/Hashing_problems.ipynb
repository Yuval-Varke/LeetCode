### 1. Top K frequent elements

```python
from collections import Counter
```

```python
def top_freq(arr,k):
    freq = Counter(arr)
    return sorted(freq,key=lambda x: freq[x], reverse = True)[:k]
```

```python
arr = [1,2,3,4,2,5,3,45,2,45,2,45,23,1,1,23,34,12,12,2]
```

```python
top_freq(arr,5)
```

```python
top_freq(arr,2)
```

### 2. Valid Anagrams

```python
def is_Anagram(s1,s2):
    if len(s1) != len(s2):
        return False

    return Counter(s1) == Counter(s2)
```

```python
inputs = [
('thing', 'night'),
('loving', 'caring'),
('listen', 'silent'),
('inch', 'chin'),
('hand', 'band')
]
```

```python
for inp in inputs:
    print(f'{inp[0]}, {inp[1]}: {is_Anagram(inp[0], inp[1])}\n')
```
