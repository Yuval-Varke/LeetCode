## Tree Traversals

### Creating the Tree

```python
class TreeNode:
    def __init__(self,val):
        self.val = val
        self.left = None
        self.right = None
```

```python
# nodes
A = TreeNode('A')
B = TreeNode('B')
C = TreeNode('C')
D = TreeNode('D')
E = TreeNode('E')
F = TreeNode('F')


# edges
A.left = B
A.right = C

B.left = D
B.right = E

C.left = F
```

### 1. Pre Order Traversal

```python
def preorder(node):
    if not node:
        return
    print(node.val,end=' ')
    preorder(node.left)
    preorder(node.right)
```

```python
preorder(A)
```

### 2. In Order Traversal

```python
def inorder(node):
    if not node:
        return
    inorder(node.left)
    print(node.val, end = ' ')
    inorder(node.right)
```

```python
inorder(A)
```

### 3. Post Order Traversal

```python
def postorder(node):
    if not node:
        return
    postorder(node.left)
    postorder(node.right)
    print(node.val,end = ' ')
```

```python
postorder(A)
```

### 4. Level Order Traversal

```python
from collections import deque
```

```python
def levelorder(node):
    if not node:
        return

    q = deque([node])

    while q:
        curr = q.popleft()
        print(curr.val, end = ' ')

        if curr.left:
            q.append(curr.left)
        if curr.right:
            q.append(curr.right)
```

```python
levelorder(A)
```
