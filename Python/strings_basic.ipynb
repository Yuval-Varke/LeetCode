### 1. Count vowels and consonents


. assume all characters are lowercase 

. string may include special characters

```python
s = '_hel lo 123$world '

v = 0
con = 0
for ch in s:
    if ch.isalpha():
        if ch in 'aeiou':
            v += 1
        else:
            con += 1
print(v)
print("---")
print(con)
```
