1. Character Count

```python
def char_cnt(text):
    freq = dict()
    for ch in text:
        freq[ch] = 1 + freq.get(ch,0)
    return freq
```

```python
char_cnt('hello')
```

```python
char_cnt('anagram is not palindrome')
```

2. First non-repeating character

```python
def first_non_repeating_char(text):
    freq = dict()
    for ch in text:
        freq[ch] = 1 + freq.get(ch,0)
    for k,v in freq.items():
        if v == 1:
            return k
    return None
```

```python
first_non_repeating_char('hello')
```

```python
first_non_repeating_char('aabbcdd')
```

```python
type(first_non_repeating_char(''))
```
