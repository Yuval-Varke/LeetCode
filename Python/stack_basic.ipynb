### Stack

```python
from collections import deque
```

1. Reverse string using stack

```python
def rev_str(s):
    stack = deque()
    result = ''

    #push into stack
    for ch in s:
        stack.append(ch)
    
    while stack:
        popped = stack.pop()
        result += popped
    
    return result
```

2. Check balanced parantheses

```python
def is_bal_par(expression) -> bool:
    stack = deque()
    pairs = {
        ')':'(',
        '}':'{',
        ']':'['
    }

    for ch in expression:
        if ch in '({[':
            stack.append(ch)
        elif ch in ')}]':
            opening_pair = pairs[ch]

            if (not stack) or opening_pair != stack.pop():
                return False
    
    return len(stack) == 0
```

```python
is_bal_par("({[]})")

# is_bal_par("")

# is_bal_par("([]})")
```
