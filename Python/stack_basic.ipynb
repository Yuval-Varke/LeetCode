### Stack

```python
from collections import deque
```

1. Reverse string using stack

```python
def rev_str(s):
    stack = deque()
    result = ''

    #push into stack
    for ch in s:
        stack.append(ch)
    
    while stack:
        popped = stack.pop()
        result += popped
    
    return result
```
