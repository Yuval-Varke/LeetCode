1. Find the largest in a list

```python
numbers = [12, 45, 7, 89, 34, 95, 62, 57]


largest = numbers[0]
for i in numbers[1:]:
    if i > largest:
        largest = i
print(f'Largest element is {largest}')
```

2. Count occurences of an element

```python
items = ["apple", "banana", "apple", "orange", "banana", "apple"]
target = "apple"

def counter(items):
    cnt = 0
    for i in items:
        if i == target:
            cnt += 1

    return cnt

counter(items)
```

3. Remove duplicates from list

```python
numbers = [1, 2, 2, 3, 4, 4, 5]

ls = []
for i in numbers:
    if i not in ls:
        ls.append(i)
print(f'Unique elements: {ls}')
```

4. Reverse a list (without using reverse method)

```python
data = [4, 6, 2, 3, 1, 7, 5]

data[::-1]
```

```python
data = [4, 6, 2, 3, 1, 7, 5]

n = len(data)
new_ls = []
for i in range(n-1,-1,-1):
    new_ls.append(data[i]) 

print(new_ls)
```
