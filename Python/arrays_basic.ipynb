1. Find the largest in a list

```python
numbers = [12, 45, 7, 89, 34, 95, 62, 57]


largest = numbers[0]
for i in numbers[1:]:
    if i > largest:
        largest = i
print(f'Largest element is {largest}')
```

2. Count occurences of an element

```python
items = ["apple", "banana", "apple", "orange", "banana", "apple"]
target = "apple"

def counter(items):
    cnt = 0
    for i in items:
        if i == target:
            cnt += 1

    return cnt

counter(items)
```
