## Binary Search Pattern Problems

### 1. First and Last occurence in sorted array

Given a sorted array arr and a target, find the first and last index

```python
def first_last_occ(arr,target):
    def first_index():
        left,right = 0,len(arr)-1
        index = -1
        while left <= right:
            mid = (left + right) // 2
            if arr[mid] >= target:
                right = mid - 1
            else:
                left = mid + 1
            if arr[mid] == target:
                index = mid
        return index

    def last_index():
        left,right = 0,len(arr)-1
        index = -1
        while left <= right:
            mid = (left + right) // 2
            if arr[mid] <= target:
                left = mid + 1
            else:
                right = mid - 1
            if arr[mid] == target:
                index = mid
        return index

    return [first_index(),last_index()]

```

```python
first_last_occ([2,4,4,4,5,7],4)
```

```python
first_last_occ([-3,-2,-2,-1,0,5,9],-2)
```

### 2. Minimum days to make Bouquets

. bloomDay[i] = day i-th flower blooms m bouquets

. Each bouquet needs k adjacent flowers

. Need to make m bouquets

```python
def min_days(bloom_days,m,k):
    if m * k > len(bloom_days):
        return -1


    def can_make(day):
        cnt = 0
        bouquets = 0

        for d in bloom_days:
            if d <= day:
                cnt += 1
            else:
                cnt = 0

            if cnt == k:
                bouquets += 1
                cnt = 0
        return bouquets >= m

    low,high  =min(bloom_days),max(bloom_days)
    res = -1
    while low <= high:
        mid = (low + high) // 2
        if can_make(mid):
            res = mid
            high = mid - 1
        else:
            low = mid + 1
    return res
```

```python
bloom_days = [1,10,3,10,2]
m=3
k=1
```

```python
min_days(bloom_days,m,k)
```

### 3. Aggresive Cows

. Given positions of stalls and number of cows

. Place cows in stalls such that minimum distance between any 2 cows is maximized

```python
def aggr_cows(stalls,cows):

    stalls.sort()

    def can_place(distance):
        placed = 1
        last = stalls[0]
        for pos in stalls[1:]:
            if pos - last >= distance:
                placed += 1
                last = pos
            if placed == cows:
                return True
        return False

    low, high = 1,stalls[-1] - stalls[0]
    res = 0

    while low <= high:
        mid = (low + high) // 2

        if can_place(mid):
            res = mid
            low = mid + 1
        else:
            high = mid - 1
    return res
    
```

```python
stalls = [1,2,8,4,9]
cows = 3

aggr_cows(stalls,cows)
```

```python
aggr_cows([1,6,4,8,2,3],1)
```
