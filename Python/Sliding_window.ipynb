## Sliding Window

For contiguous arrays or strings

Traditional = worse = O(n^2)

Sliding window = O(n)

### 1. Maximum sum subarray of size K   

Given an array arr and an integer k, find the maximum sum of any contiguous subarray of size
k

```python
def max_sum_sub(arr,k):
    max_sum = 0
    for i in range(k):
        max_sum += arr[i]
    sum = max_sum
    for i in range(k,len(arr)):
        sum = sum - arr[i-k]
        sum = sum + arr[i]
        if sum > max_sum:
            max_sum = sum
    return max_sum
```

```python
arr = [1,3,5,6,2,8,1,9]
```

```python
max_sum_sub(arr,2)
```

```python
max_sum_sub([1,2,3,4,5,6,7,8,9],3)
```

### 2. Longest Substring without repeating characters

Given a string s, return the length of the longest substring with no repeating characters

```python
def length_substring(s):
    char_set = set()
    max_len = 0
    left = 0

    for right in range(len(s)):
        while s[right] in char_set:
            char_set . remove (s[left])
            left += 1
        char_set. add(s[right])
        max_len = max(max_len, right - left + 1)

    return max_len
```

```python
length_substring('yuval')
```

```python
length_substring('radar')
```

```python
length_substring('abbcdeafddcbcbba')
```
