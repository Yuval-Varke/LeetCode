## Binary Search Trees

```python
class TreeNode:
    def __init__(self, val):
        self.val = val
        self.left = None
        self.right = None
```

```python
root = TreeNode(40)
root.left = TreeNode(30)
root.right = TreeNode(50)

root.left.left = TreeNode(25)
root.left.right = TreeNode(35)
root.right.left = TreeNode(45)
root.right.right = TreeNode(60)
```

```python
def inorder(node):
    if not node:
        return
        
    inorder(node.left)
    print(node.val,end=" ")
    inorder(node.right)
```

```python
inorder(root)
```
