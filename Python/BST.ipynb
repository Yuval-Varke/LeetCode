## Binary Search Trees

```python
class TreeNode:
    def __init__(self, val):
        self.val = val
        self.left = None
        self.right = None
```

```python
root = TreeNode(40)
root.left = TreeNode(30)
root.right = TreeNode(50)

root.left.left = TreeNode(25)
# root.left.right = TreeNode(35)
root.right.left = TreeNode(45)
root.right.right = TreeNode(60)
```

```python
def inorder(node):
    if not node:
        return
        
    inorder(node.left)
    print(node.val,end=" ")
    inorder(node.right)
```

```python
inorder(root)
```

### 1. Insertion at BST

```python
def insert(root, key):
    if not root:
        return TreeNode(key)

    if key < root.val:
        root.left = insert(root.left,key)
    else:
        root.right = insert(root.right,key)

    return root
```

```python
insert(root,35)
```

```python
inorder(root)
```

### 2. Searching at BST

```python
def search(node,target):
    if not node or (node.val == target):
        return node

    if target < node.val:
        return search(node.left,target)
    
    return search(node.right,target)
```

```python
node = search(root,35)
```
