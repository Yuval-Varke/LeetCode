## N-ary Tree

### 1. Setting up N ary Tree

```python
class Node:
    def __init__(self, val):
        self.val = val
        self.children = [] 
```

```python
root = Node(10)

# level 1
node_20 = Node(20)
node_30 = Node(30)
node_40 = Node(40)

root.children = [node_20, node_30, node_40]

# level 2
node_20.children = [Node(50), Node(60), Node(70)]
node_30.children = [Node(80)]
node_40.children = [Node(90), Node(100)]
```

### 2. Pre Order Traversal

```python
def preorder(node):
    if not node:
        return None
    print(node.val,end=' ')
    for child in node.children:
        preorder(child)
```

```python
preorder(root)
```

### 3. Post Order Traversal

```python
def postorder(node):
    if not node:
        return None
    for child in node.children:
        postorder(child)
    print(node.val,end=' ')
```

```python
postorder(root)
```
