## Prefix Sum

###### Important ====> Sum(arr[i...j]) = P[j+1] - P[i]

### 1. Subarray sum equals K 

```python
def sub_array_sum(arr, k):
    count = 0
    prefix_sum = 0
    freq = {0: 1}

    for num in arr:
        prefix_sum += num
        if prefix_sum - k in freq:
            count += freq[prefix_sum - k]
        freq[prefix_sum] = freq.get(prefix_sum, 0) + 1

    return count
```

```python
sub_array_sum([1,2,3],3)
```

```python
sub_array_sum([1,1,1],2)
```

### 2. Count Subarrays divisible by k

```python
def count_sub_arrays(arr,k):
    cnt = 0
    freq = {0:1}
    p_sum = 0

    for num in arr:
        p_sum += num
        mod = p_sum % k
        if mod in freq:
            cnt += freq[mod]
        freq[mod] = freq.get(mod,0) + 1
    return cnt
```

```python
count_sub_arrays([4,5,0,-2,-3,1],5)
```

```python
count_sub_arrays([0,0,0],3)
```

```python
count_sub_arrays([1,2,3],7)
```

```python
count_sub_arrays([5],5)
```
